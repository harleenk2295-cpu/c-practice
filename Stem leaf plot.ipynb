```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv("/home/student/2025510071/winequality-red.csv",sep=",")

def stem_and_leaf(data,leaf_unit=1):
    stems = np.floor(data / leaf_unit).astype(int)
    leaves = (data % leaf_unit * 10).astype(int)
    return stems,leaves
stem,leaf = stem_and_leaf(df["fixed acidity"])

plt.stem(stem,leaf)
plt.xlabel("Stem")
plt.ylabel("Leaf")
plt.title("fixed acidity")
plt.show()
```
