```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_csv("C:/Users/Bathm/Downloads/archive (5)/WineQT.csv", sep=",")
def stem_and_leaf(data, leaf_unit=1):
    stems = np.floor(data / leaf_unit).astype(int)
    leafs = ((data % leaf_unit) * 10).astype(int)
    return stems, leafs
stems, leafs = stem_and_leaf(df["fixed acidity"])
plt.stem(stems, leafs)
plt.xlabel("Stem")
plt.ylabel("Leaf")
plt.title("Stem and Leaf Plot - Fixed Acidity")
plt.show()
```
