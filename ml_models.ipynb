```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.cluster import KMeans
from sklearn.metrics import accuracy_score, mean_squared_error
from sklearn.preprocessing import LabelEncoder, StandardScaler
from scipy.stats import mode
# Load the Iris dataset
iris = pd.read_csv('/home/student/2025510071/iris.csv')
# Encode the target variable
m = LabelEncoder()
iris['species'] = m.fit_transform(iris['species'])
X = iris.drop('species', axis=1) # Drop the target column from features
y = iris['species']
# Normalize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
# Split the dataset into training and test sets
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
# Initialize the models
models = {
    "SVC": SVC(),
    "KNN": KNeighborsClassifier(),
    "Random Forest": RandomForestClassifier(),
    "Logistic Regression": LogisticRegression(max_iter=200),
    "Naive Bayes": GaussianNB(),
    "Linear Regression": LinearRegression()
}
# Fit and evaluate each model
accuracies = {}
for model_name, model in models.items():
    model.fit(X_train, y_train)
    predictions = model.predict(X_test)
    if model_name == "Linear Regression":
# Convert predictions to class labels (rounding)
        predicted_classes = np.round(predictions).astype(int)
# Ensure predictions are within valid class range [0, 2]
        predicted_classes = np.clip(predicted_classes, 0, 2)
        accuracies[model_name] = accuracy_score(y_test, predicted_classes) * 100
    else:
        accuracies[model_name] = accuracy_score(y_test, predictions) * 100
# K-means clustering (unsupervised)
kmeans = KMeans(n_clusters=3, random_state=42,n_init=10)
kmeans.fit(X_scaled) # Fit K-means on the scaled features
kmeans_labels = kmeans.labels_
# Map K-means labels to true labels
def get_accuracy_from_kmeans(true_labels, kmeans_labels):
    label_mapping = {}
    for cluster in np.unique(kmeans_labels):
        true_labels_in_cluster = true_labels[kmeans_labels == cluster]
        most_common_label = mode(true_labels_in_cluster,keepdims=False).mode
        label_mapping[cluster] = most_common_label
        mapped_labels = np.array([label_mapping[cluster] for cluster in kmeans_labels])
        return accuracy_score(true_labels, mapped_labels) * 100
# Calculate accuracy for K-means
kmeans_accuracy = get_accuracy_from_kmeans(y.to_numpy(), kmeans_labels)

# Print the K-means accuracy
print(f"K-means Accuracy: {kmeans_accuracy:.2f}")
# Print the accuracies of the supervised models
for model_name, score in accuracies.items():
    print(f"{model_name} Accuracy: {score:.2f}")
```

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.cluster import KMeans
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler
from collections import Counter

iris = pd.read_csv('/home/student/2025510071/heart.csv')
X = iris.drop('target', axis=1)
y = iris['target']
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
models = {
    "SVC": SVC(),
    "KNN": KNeighborsClassifier(),
    "Random Forest": RandomForestClassifier(random_state=42),
    "Logistic Regression": LogisticRegression(max_iter=200),
    "Naive Bayes": GaussianNB(),
    "Linear Regression": LinearRegression()
}
accuracies = {}
for model_name, model in models.items():
    model.fit(X_train, y_train)
    predictions = model.predict(X_test)
    if model_name == "Linear Regression":
        predicted_classes = np.round(predictions).astype(int)
        min_class = int(y.min())
        max_class = int(y.max())
        predicted_classes = np.clip(
            predicted_classes,
            min_class,
            max_class
        )

        accuracies[model_name] = (
            accuracy_score(y_test, predicted_classes) * 100
        )
    else:

        accuracies[model_name] = (
            accuracy_score(y_test, predictions) * 100
        )
n_clusters = y.nunique()
kmeans = KMeans(
    n_clusters=n_clusters,
    random_state=42,
    n_init=10
)
kmeans.fit(X_scaled)
kmeans_labels = kmeans.labels_
def get_accuracy_from_kmeans(true_labels, kmeans_labels):
    true_labels = np.asarray(true_labels)
    kmeans_labels = np.asarray(kmeans_labels)
    label_mapping = {}
    for cluster in np.unique(kmeans_labels):
        true_labels_in_cluster = true_labels[
            kmeans_labels == cluster
        ]
        most_common_label = Counter(
            true_labels_in_cluster
        ).most_common(1)[0][0]
        label_mapping[cluster] = most_common_label
    mapped_labels = np.array([
        label_mapping[cluster]
        for cluster in kmeans_labels
    ])

    return accuracy_score(
        true_labels,
        mapped_labels
    ) * 10
kmeans_accuracy = get_accuracy_from_kmeans(
    y,
    kmeans_labels
)
print(f"K-Means Accuracy: {kmeans_accuracy:.2f}%")
print("\nSupervised Model Accuracies:")
for model_name, score in accuracies.items():
    print(f"{model_name}: {score:.2f}%")
```
