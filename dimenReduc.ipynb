```python
import pandas as pd
import seaborn as sns
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn import metrics
df=pd.read_csv('/home/student/2025510071/heart.csv')
a=df.drop(['sex'],axis=1)
b=df['sex']
scaling=StandardScaler()
scaling.fit(a)
Scaled_data=scaling.transform(a)
ori=pd.DataFrame(Scaled_data)
X_train, X_test, y_train, y_test = train_test_split(Scaled_data, b, test_size=0.60,random_state=150)
clf = svm.SVC(kernel='linear')
clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)
ac3=metrics.accuracy_score(y_test, y_pred)*100
print("Accuracy:",ac3)
```

```python
i=1
acc=[]
name=[]
while i<4:
    pca1=PCA(n_components=i)
    pca1.fit(Scaled_data)
    x=pca1.transform(Scaled_data)
    pca_1=pd.DataFrame(x)
    X_train, X_test, y_train, y_test = train_test_split(pca_1, b, test_size=0.60,random_state=150)
    clf = svm.SVC(kernel='linear')
    clf.fit(X_train, y_train)
    y_pred = clf.predict(X_test)
    acc.append(metrics.accuracy_score(y_test, y_pred)*100)
    name.append('Pca {}'.format(i))
    print('pca {} accuracy:'.format(i),acc[i-1])
    i=i+1
```
