```python
import pandas as pd

df = pd.read_csv("/home/student/2025510071/ucidiabetes(1).csv")

for i in df.index:
    if df.loc[i, 'value'] in ['0Hi', '0Lo', '3A', "0''"]:
        df.drop(i, inplace=True)

df.describe()
```

```python
df['value'] = pd.to_numeric(df['value'], errors='coerce')

mean_value = df.groupby('patient_id')['value'].mean()

print(mean_value)
```

```python
FRE=df.groupby('patient_id')['value'].value_counts()
print(FRE)
```
